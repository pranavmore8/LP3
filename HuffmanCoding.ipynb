```python
# Huffman Encoding
import heapq

class Node:
    def __init__(self, ch, freq, left=None, right=None):
        self.ch, self.freq, self.left, self.right = ch, freq, left, right
    def __lt__(self, other):
        return self.freq < other.freq

def build_codes(node, code, codes):
    if node is None:
        return
    if node.ch is not None:
        codes[node.ch] = code or "0"
        return
    build_codes(node.left, code + "0", codes)
    build_codes(node.right, code + "1", codes)

text = input("Enter text: ")

freq = {}
for c in text:
    freq[c] = freq.get(c, 0) + 1

heap = [Node(c, f) for c, f in freq.items()]
heapq.heapify(heap)

while len(heap) > 1:
    a = heapq.heappop(heap)
    b = heapq.heappop(heap)
    heapq.heappush(heap, Node(None, a.freq + b.freq, a, b))

codes = {}
build_codes(heap[0], "", codes)

print("Character  Frequency  Code")
for c in codes:
    print("%-10r %-10d %s" % (c, freq[c], codes[c]))

encoded = "".join(codes[c] for c in text)
print("Encoded:", encoded)
print("Original bits:", len(text) * 8, "| Encoded bits:", len(encoded))
```

```python
# Job Sequencing with Deadlines
jobs = []
n = int(input("Enter number of jobs: "))
for i in range(n):
    jid = input("Job id: ")
    d = int(input("Deadline: "))
    p = int(input("Profit: "))
    jobs.append((jid, d, p))

jobs.sort(key=lambda x: x[2], reverse=True)

max_d = max(j[1] for j in jobs)
slot = [None] * (max_d + 1)
total = 0

for jid, d, p in jobs:
    for t in range(min(d, max_d), 0, -1):
        if slot[t] is None:
            slot[t] = jid
            total += p
            break

print("Job sequence:", [s for s in slot[1:] if s])
print("Total profit:", total)
```
