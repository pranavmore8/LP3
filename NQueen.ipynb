```python
N = 8
board = [[0] * N for _ in range(N)]

def safe(r, c):
    for i in range(N):
        if board[r][i] == 1 or board[i][c] == 1:
            return False
    for i in range(N):
        for j in range(N):
            if board[i][j] == 1 and abs(i - r) == abs(j - c):
                return False
    return True

def solve(col):
    if col == N:
        return True
    for r in range(N):
        if safe(r, col):
            board[r][col] = 1
            if solve(col + 1):
                return True
            board[r][col] = 0
    return False

r = int(input("Enter row (0-7) for first queen in column 0: "))
board[r][0] = 1

if solve(1):
    for row in board:
        print(*row)
else:
    print("No solution with this starting position")
```
